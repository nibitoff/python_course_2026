# Часть 1. Работа со строками и исключениями

Функция `parse_int` удаляет лишние пробелы методом `strip()`, затем преобразовывает строку в целое число. Если преобразование невозможно, перехватывается исключение `ValueError`, и функция возвращает `None`.

```python
def parse_int(s: str) -> int | None:
    s = s.strip()
    try:
        return int(s)
    except ValueError:
        return None


examples = ["12", " -7 ", "12a", ""]

for example in examples:
    print(f"parse_int({example!r}) -> {parse_int(example)}")
```

# Часть 2. Циклы и работа со списками

Создаем список из 25 целых чисел. Далее реализуем три функции с использованием циклов `for`.

```python
nums = [12, -5, 7, 0, 18, -11, 3, 24, 9, -2,
        15, 6, -8, 21, 14, -1, 30, 5, 2, -16,
        11, 27, 4, -6, 19]

print("Количество элементов:", len(nums))
print("Список:", nums)
```

## 2.1. Функция `count_positive`

Функция проходит по всем элементам списка и увеличивает счётчик, если текущее число больше нуля.

```python
def count_positive(nums: list[int]) -> int:
    count = 0
    for num in nums:
        if num > 0:
            count += 1
    return count


print("Количество положительных чисел:", count_positive(nums))
```

## 2.2. Функция `sum_even`

Функция Возвращает сумму всех чётных чисел.


```python
def sum_even(nums: list[int]) -> int:
    total = 0
    for num in nums:
        if num % 2 == 0:
            total += num
    return total


print("Сумма чётных чисел:", sum_even(nums))
```

## 2.3. Функция `first_divisible`

Функция ищет первое число, которое делится на `k`. После нахождения подходящего числа цикл прекращается с помощью `break`. Если подходящего числа нет, возвращается `None`.

```python
def first_divisible(nums: list[int], k: int) -> int | None:
    result = None

    for num in nums:
        if num % k == 0:
            result = num
            break

    return result


print("Первое число, делящееся на 5:", first_divisible(nums, 5))
print("Первое число, делящееся на 13:", first_divisible(nums, 13))
```

# Часть 3. Lambda и сортировки

Для сортировки используем функцию `sorted()` и задаём правило сортировки через `key=lambda ...`.

```python
items = [
    {"name": "A", "price": 120, "rating": 4.7},
    {"name": "B", "price": 90, "rating": 4.2},
    {"name": "C", "price": 150, "rating": 4.9},
]

items_by_price = sorted(items, key=lambda item: item["price"])

print("Сортировка по цене:")
for item in items_by_price:
    print(item)

items_by_rating = sorted(items, key=lambda item: item["rating"], reverse=True)

print("Сортировка по рейтингу:")
for item in items_by_rating:
    print(item)
```

# Часть 4. Мини-ООП

Создаем класс `Chocolate`. Объект хранит название шоколада и его цену. Если цена меньше или равна нулю, выдаем `ValueError`. Метод `discount` рассчитывает новую цену после применения скидки.

```python
class Chocolate:
    def __init__(self, name: str, price: float):
        if price <= 0:
            raise ValueError("Цена должна быть больше нуля")

        self.name = name
        self.price = price

    def discount(self, percent: int) -> float:
        return self.price * (1 - percent / 100)
```

Создадим два объекта класса и проверим работу метода `discount`.

```python
chocolate_1 = Chocolate("Milk Chocolate", 250.0)
chocolate_2 = Chocolate("Dark Chocolate", 320.0)

print(chocolate_1.name, "со скидкой 10%:", chocolate_1.discount(10))
print(chocolate_2.name, "со скидкой 25%:", chocolate_2.discount(25))
```
